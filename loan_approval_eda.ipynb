### Univariate Analysis

```python
df.head()
```

```python
sns.countplot(x = df.Gender)
```

```python
categorical = []

for i in df:
    if len(df[i].unique())<=20:
        categorical.append(i)
```

```python
categorical
```

```python
df_cat = df[categorical]
```

```python
continuous = []

for i in df:
    if len(df[i].unique())>=20:
        continuous.append(i)
```

```python
df_cont = df[continuous]
```

```python
df_cont.drop(columns = 'Loan_ID',inplace =True) 

# or df.cont = df_cont.drop(column = 'Loan_ID')
```

```python
plotnumber = 1

plt.figure(figsize =(10,10))
for i in df_cat:
    plt.subplot(3, 3 ,plotnumber)
    sns.countplot(x = df_cat[i])
    plt.xticks(rotation = 45)
    plotnumber +=1
plt.tight_layout()
```

## Insights

1)Gender
In gender column there are total 489 male applicant and 112 female applicant
so comparitively there are more Number of Male applicants than females i.e 81% of Male and 19% female.

2)In this Loan data there are more married people than non-married,there are approximately 65% of Married
people and 35% of Non Married people.

3)As the above graph of dependents shows us how there are more people with 0 dependents than others, i.e 
there are around 345 people with 0 dependents,102 people with 1 dependents,101 with 2 dependents and 51 
people with 3+ dependents in family 

4)The fourth chart show us the number of Graduate and non-Graduate people and clearly the percentage of graduate
are far more than non graduate people that is 78:22

5)Most applicant are not self_employed roughly 490 compared with about 70 self_employed applicants

6)Loan Amount Term shows most of the loan amount tenure is about 360 months that is about 30 years 
this dominates other loan repayment tenures, also it says that the applicants are more relaible & comfortable to 30 years 
tenure of loan repayment than other tenures

7)Semiurban applicants group appears to be largest group after Urban and then rural applicants

8)There are approximately 422 approved(Y) and 192 rejected(N) application,this clearly indicates the loan rejected are far 
smallar group than loan approved.

So the loan data shows the majority of the applicant are male,graduates,non_self employed and have positive credit score.
Most of the loan tenure are 360 months i.e 30 years and larger no of applicants are from semiurban area.
